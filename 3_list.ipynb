```python
#to remove whole list
list1 = [100, "hello", 12.21, "abhinav", True,1,2,3]
print(id(list1))  #where list is present in memory

list1.clear()

print(id(list1))  #same memory location but list is empty
print(list1)

#here we are clearing previous list only so same address
```

```python
list1 = [1,2,3]
list2 = list1 # deep copy

list1[0] = 100
list2[1] = 100

print(list1,id(list1))
print(list2,id(list2))
```

```python
list1 = [1,2,3]
list2 = list1.copy() # shallow copy

list1[0] = 100
list2[1] = 100

print(list1,id(list1))
print(list2,id(list2))
```
